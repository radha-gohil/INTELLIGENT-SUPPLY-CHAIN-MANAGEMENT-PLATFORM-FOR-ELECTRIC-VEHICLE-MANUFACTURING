```python
supplier_network_df = supplier_component_df.merge(
    supplier_master_df,
    on="supplier_id",
    how="left",
    suffixes=("_component", "_supplier")
)

display(supplier_network_df.head())
```

```python
import pandas as pd
import numpy as np

np.random.seed(42)

# Number of purchase orders
N = 2000

# Available suppliers/components
supplier_ids = supplier_component_df["supplier_id"].unique()
part_ids = supplier_component_df["part_id"].unique()

warehouses = [
    "Chennai",
    "Hosur",
    "Coimbatore",
    "Chengalpattu"
]

po_data = []

for i in range(N):

    # Select supplier-component relationship
    relationship = supplier_component_df.sample(1).iloc[0]

    supplier_id = relationship["supplier_id"]
    part_id = relationship["part_id"]

    # Supplier-specific information
    lead_time = int(relationship["lead_time_days"])
    unit_cost = float(relationship["unit_cost"])

    # PO date
    po_date = pd.Timestamp(
        np.random.choice(
            pd.date_range(
                "2025-01-01",
                "2026-08-31"
            )
        )
    )

    # Ordered quantity
    quantity_ordered = int(
        np.random.randint(
            100,
            2000
        )
    )

    # Random delivery behavior
    delivery_behavior = np.random.choice(
        ["On Time", "Late", "Partial", "Late + Partial"],
        p=[0.60, 0.20, 0.12, 0.08]
    )

    # Expected delivery
    expected_date = (
        po_date +
        pd.Timedelta(days=lead_time)
    )

    # Actual delivery
    if delivery_behavior == "On Time":

        delay = np.random.randint(-2, 3)

        actual_date = (
            expected_date +
            pd.Timedelta(days=delay)
        )

        quantity_received = quantity_ordered

        status = "Delivered"

    elif delivery_behavior == "Late":

        delay = np.random.randint(3, 15)

        actual_date = (
            expected_date +
            pd.Timedelta(days=delay)
        )

        quantity_received = quantity_ordered

        status = "Late"

    elif delivery_behavior == "Partial":

        delay = np.random.randint(-1, 4)

        actual_date = (
            expected_date +
            pd.Timedelta(days=delay)
        )

        quantity_received = int(
            quantity_ordered *
            np.random.uniform(0.60, 0.95)
        )

        status = "Partial"

    else:

        delay = np.random.randint(3, 15)

        actual_date = (
            expected_date +
            pd.Timedelta(days=delay)
        )

        quantity_received = int(
            quantity_ordered *
            np.random.uniform(0.50, 0.90)
        )

        status = "Late + Partial"

    warehouse = np.random.choice(warehouses)

    po_data.append([
        f"PO{i+1:05d}",
        po_date,
        supplier_id,
        part_id,
        warehouse,
        quantity_ordered,
        quantity_received,
        unit_cost,
        expected_date,
        actual_date,
        status
    ])
```

```python
purchase_order_df = pd.DataFrame(
    po_data,
    columns=[
        "po_id",
        "po_date",
        "supplier_id",
        "part_id",
        "warehouse",
        "quantity_ordered",
        "quantity_received",
        "unit_cost",
        "expected_delivery_date",
        "actual_delivery_date",
        "order_status"
    ]
)

print("Shape:", purchase_order_df.shape)

display(
    purchase_order_df.head(20)
)
```

```python
purchase_order_df["quantity_shortage"] = (
    purchase_order_df["quantity_ordered"]
    -
    purchase_order_df["quantity_received"]
)
```

```python
purchase_order_df["delivery_delay_days"] = (
    purchase_order_df["actual_delivery_date"]
    -
    purchase_order_df["expected_delivery_date"]
).dt.days
```

```python
purchase_order_df["order_value"] = (
    purchase_order_df["quantity_ordered"]
    *
    purchase_order_df["unit_cost"]
)
```

```python
display(
    purchase_order_df.head()
)
```

```python
supplier_performance = (
    purchase_order_df
    .groupby("supplier_id")
    .agg(
        total_orders=("po_id", "count"),
        total_ordered=("quantity_ordered", "sum"),
        total_received=("quantity_received", "sum"),
        average_delay_days=("delivery_delay_days", "mean"),
        total_shortage=("quantity_shortage", "sum")
    )
    .reset_index()
)

display(
    supplier_performance.sort_values(
        "average_delay_days",
        ascending=False
    )
)
```

```python
supplier_performance["fulfillment_rate"] = (
    supplier_performance["total_received"]
    /
    supplier_performance["total_ordered"]
) * 100
```

```python
display(
    supplier_performance.sort_values(
        "fulfillment_rate"
    )
)
```
