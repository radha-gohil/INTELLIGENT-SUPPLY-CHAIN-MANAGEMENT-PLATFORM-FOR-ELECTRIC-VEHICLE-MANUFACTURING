```python
import pandas as pd
import numpy as np

np.random.seed(42)

warehouses = [
    "Chennai",
    "Hosur",
    "Coimbatore",
    "Chengalpattu"
]

inventory_data = []

for _, part in component_master_df.iterrows():

    for warehouse in warehouses:

        part_id = part["part_id"]
        part_name = part["part_name"]

        # Different stock levels for different components
        available_stock = np.random.randint(500, 10000)

        # Safety stock
        safety_stock = int(
            available_stock * np.random.uniform(0.10, 0.25)
        )

        # Reorder level should be higher than safety stock
        reorder_level = int(
            safety_stock * np.random.uniform(1.5, 2.5)
        )

        # Determine inventory status
        if available_stock <= safety_stock:
            inventory_status = "Critical"

        elif available_stock <= reorder_level:
            inventory_status = "Reorder"

        else:
            inventory_status = "Healthy"

        inventory_data.append([
            part_id,
            part_name,
            warehouse,
            available_stock,
            safety_stock,
            reorder_level,
            inventory_status
        ])


inventory_df = pd.DataFrame(
    inventory_data,
    columns=[
        "part_id",
        "part_name",
        "warehouse",
        "available_stock",
        "safety_stock",
        "reorder_level",
        "inventory_status"
    ]
)

print("Shape:", inventory_df.shape)

display(inventory_df.head(20))
```

```python
reorder_items = inventory_df[
    inventory_df["inventory_status"].isin(
        ["Reorder", "Critical"]
    )
]

display(
    reorder_items.sort_values(
        "available_stock"
    )
)
```
